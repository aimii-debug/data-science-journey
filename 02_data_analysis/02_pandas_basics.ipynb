# Pandas Basics Practice Portfolio

This notebook contains practical Pandas exercises and a portfolio-style mini project for data analysis. Use it to build your Pandas portfolio with real-world data skills.

```python
import pandas as pd
import numpy as np

print('Pandas version:', pd.__version__)
print('NumPy version:', np.__version__)
```

## 1. DataFrame Creation

Create DataFrames from dictionaries, lists, and CSV-like data. This is the foundation of any Pandas portfolio project.

```python
sales_data = {
    'Date': ['2024-01-05', '2024-01-07', '2024-01-10', '2024-01-12', '2024-01-15', '2024-01-17'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South'],
    'Product': ['Laptop', 'Tablet', 'Phone', 'Laptop', 'Headphones', 'Phone'],
    'Quantity': [5, 8, 12, 3, 20, 7],
    'Unit Price': [1200, 450, 700, 1300, 120, 680],
    'Profit': [1500, 900, 2100, 500, 1400, 850],
    'Customer Segment': ['Corporate', 'Consumer', 'Consumer', 'Corporate', 'Consumer', 'Corporate']
}
df = pd.DataFrame(sales_data)
df['Date'] = pd.to_datetime(df['Date'])
df['Sales'] = df['Quantity'] * df['Unit Price']
df
```

## 2. Inspecting the Data

Use these methods to understand the structure and summary of your dataset. Every portfolio analysis begins here.

```python
print('Shape:', df.shape)
print('
Columns:', df.columns.tolist())
print('
Info:')
df.info()
print('
Describe:')
df.describe(include='all')
```

## 3. Selection and Indexing

Learn how to select rows and columns using `loc`, `iloc`, and boolean filtering.

```python
# Select a single column
print(df['Product'])

# Select multiple columns
print(df[['Date', 'Region', 'Sales']])

# Select rows by label
print(df.loc[0:2, ['Date', 'Product', 'Sales']])

# Select rows by integer position
print(df.iloc[1:4, 0:4])

# Boolean filtering
print(df[df['Sales'] > 2000])
```

## 4. Data Cleaning

Cleaning is essential for portfolio-ready projects. Practice handling missing values, duplicates, and type conversions.

```python
# Simulate missing values and duplicates
df_clean = df.copy()
df_clean.loc[2, 'Profit'] = np.nan
df_clean.loc[4, 'Region'] = None
df_clean = df_clean.append(df_clean.iloc[0], ignore_index=True)
print('Before cleaning:')
print(df_clean)

print('Missing values:')
print(df_clean.isna().sum())

# Fill and drop missing values
filled = df_clean.fillna({'Region': 'Unknown', 'Profit': df_clean['Profit'].mean()})
print('
Filled missing values:')
print(filled)

# Drop duplicates
no_duplicates = filled.drop_duplicates()
print('
After dropping duplicates:')
print(no_duplicates)
```

## 5. Grouping and Aggregation

Use groupby to summarize sales by region, product, and customer segment. This is key for portfolio storytelling.

```python
print('Sales by region:')
print(df.groupby('Region')[['Sales', 'Profit']].sum())

print('Average sales by product:')
print(df.groupby('Product')['Sales'].mean())

print('Portfolio summary by customer segment:')
print(df.groupby('Customer Segment').agg({'Sales': 'sum', 'Profit': 'sum', 'Quantity': 'mean'}))
```

## 6. Merging and Joining

Practice combining datasets, which is an important skill for portfolio projects that use multiple sources.

```python
customer_data = pd.DataFrame({
    'Customer Segment': ['Corporate', 'Consumer'],
    'Priority': ['High', 'Medium']
})
merged = df.merge(customer_data, on='Customer Segment', how='left')
print('Merged dataset:')
print(merged)
```

## 7. Portfolio Project: Sales Performance Dashboard

Build a mini portfolio project by analyzing product and regional performance using Pandas.

```python
monthly_summary = df.copy()
monthly_summary['Month'] = monthly_summary['Date'].dt.to_period('M')
summary = monthly_summary.groupby(['Month', 'Region']).agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum'
}).reset_index()
print('Monthly sales summary by region:')
print(summary)

top_products = df.groupby('Product')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
print('
Top products by sales:')
print(top_products)
```

## 8. Practice Exercises

Try these exercises to strengthen your portfolio-ready Pandas skills.

1. Create a new column named `Profit Margin` as `Profit / Sales`.
2. Find the top-selling product by quantity.
3. Filter transactions with `Sales` greater than 2000.
4. Calculate total sales for the `Consumer` segment.
5. Sort the DataFrame by `Date` and show the first 5 rows.

```python
# Exercise solutions
df_ex = df.copy()
df_ex['Profit Margin'] = df_ex['Profit'] / df_ex['Sales']
print('Profit Margin added:')
print(df_ex[['Product', 'Sales', 'Profit', 'Profit Margin']])

print('
Top-selling product by quantity:')
print(df_ex.groupby('Product')['Quantity'].sum().sort_values(ascending=False).head(1))

print('
Transactions with Sales > 2000:')
print(df_ex[df_ex['Sales'] > 2000])

print('
Total sales for Consumer segment:')
print(df_ex[df_ex['Customer Segment'] == 'Consumer']['Sales'].sum())

print('
Sorted by Date:')
print(df_ex.sort_values('Date').head())
```
