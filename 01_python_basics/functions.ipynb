```python
#Write a function that returns the square of a number
def sqr(a):
    c=a**2
    return c

print(sqr(2))
print(sqr(6))
print(sqr(4))
```

```python
#Write a function that returns True if the number is prime, else False
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
        
is_prime(17)
```

```python
#Write a function to calculate factorial using a loop
def fac(n):
    c=1
    for i in range(1,n+1):
        c*=i
    return c
        
fac(8)        
```
