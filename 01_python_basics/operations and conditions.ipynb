```python
#Ask the user for a number and print if it’s even or odd.
u=int(input("Enter a number: "))
if u%2==0:
    print(f"You entered {u} and The number is even")
else:
    print(f"You entered {u} and The number is odd")

'''Ask for a score (0–100) and print grade:
A: 90–100
B: 80–89
C: 70–79
F: below 70'''

v=int(input("Enter your score: "))
if v>=90:
    print(f"Your score is {v} and Your grade is A")
if v>=80 and v<90:
    print(f"Your score is {v} and Your grade is B")
if v>=70 and v<80:
    print(f"Your score is {v} and Your grade is C")
if v<70:
    print(f"Your score is {v} and Your grade is F")

```

```python
#Check if a number is divisible by 3 and 5
c=int(input("Enter number to check divisibility: "))
if c%3==0 and c%5==0:
    print(f"{c} is divisible by both 3 and 5")
elif c%3==0:
    print(f"{c} is divisible by 3")
elif c%5==0:
    print(f"{c} is divisible by 5")
else:
    print(f"{c} is neither divisible by 3 nor 5")
```
